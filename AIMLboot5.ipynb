```python
#step:1

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
print("Libraries imported successfully")
print("\n")
#step2

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
print("\n")

#step3

plt.imshow(
    x_train[0],
    cmap="gray"
)

plt.title("First MNIST Image")

plt.axis("off")

plt.show()
print("\n")

print(y_train[0])
print("\n")

#step4

x_train = x_train.astype("float32") / 255.0

x_test = x_test.astype("float32") / 255.0

#step5

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28,28)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(
        128,
        activation="relu"
    ),
    tf.keras.layers.Dense(
        10,
        activation="softmax"
    )
])
tf.keras.layers.Input(
    shape=(28,28)
)
tf.keras.layers.Flatten()
tf.keras.layers.Dense(
    128,
    activation="relu"
)
tf.keras.layers.Dense(
    10,
    activation="softmax"
)
#step:6

model.summary()

#step:7

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
#step:8

history = model.fit(
    x_train,
    y_train,
    epochs=5
)

#step:9

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test
)

print(test_loss)
print(test_accuracy)
print("\n")
 #step:10

prediction = model.predict(
    x_test[0:1],
    verbose=0
)

#step:11

for digit in range(10):

    print(
        prediction[0][digit]
    )
#step:12

predicted_digit = tf.argmax(
    prediction[0]
).numpy()


#step:13

print(y_test[0])

print(predicted_digit)
print("\n")
```
