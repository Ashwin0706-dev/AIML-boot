```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx
from sklearn.cluster import SpectralClustering
from sklearn.metrics import silhouette_score
from sklearn.metrics.pairwise import cosine_similarity
df=pd.read_csv('social_network.csv')
print("original Dataset:")
display(df)

students=df["Student"]
X=df.drop("Student",axis=1)
print("\nStudent Names:")
print(students.tolist())
print("\nFeature Matrix:")
display(X)

A=cosine_similarity(X)
print("\nCosine Similarity Matrix:")
print(np.round(A,3))

degree=A.sum(axis=1)
print("\nDegree of each Student:")
for student,d in zip(students,degree):
  print(student,":",round(d,3))

D=np.diag(degree)
print("\nDegree Matrix:")
print(np.round(D,3))

L=D-A
print("\nLaplacian Matrix:")
print(np.round(L,3))

n_clusters=2
model=SpectralClustering(
    n_clusters=n_clusters,
    affinity="precomputed",
    random_state=42
)
labels=model.fit_predict(A)

result=pd.DataFrame({
    "Student":students,
    "Cluster":labels
})
result["Cluster"]=result["Cluster"]+1
print("\nCluster Results:")
display(result)

silhouette_labels = result["Cluster"] - 1

score = silhouette_score(
    X,
    silhouette_labels
)

print("\nSilhouette Score:", round(score, 4))
```
